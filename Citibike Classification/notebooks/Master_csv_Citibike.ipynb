### Formatting File to combine the excel files into one dataset for better processing 

```python
import pandas as pd
import glob
```

```python
file_list = glob.glob("*.csv")
df_list = [pd.read_csv(file) for file in file_list]
```

```python
master_df = pd.concat(df_list, axis=0, ignore_index=True)
master_df.to_csv("master_file.csv", index=False)

print(f"Successfully combined {len(file_list)} files!")
```
